# 📊 Medical Cost Personal Dataset


## 🌍 Problématique globale

Dans un contexte où les dépenses de santé ne cessent d’augmenter, les compagnies d’assurance et les systèmes de santé cherchent à mieux comprendre les facteurs qui influencent les coûts médicaux. La question centrale est la suivante : **peut‑on prédire de manière fiable les charges médicales individuelles à partir de caractéristiques socio‑démographiques et comportementales ?**  
Ce problème est crucial car il permet non seulement d’anticiper les dépenses et d’ajuster les politiques tarifaires, mais aussi de mettre en évidence l’impact de variables comme l’âge, l’indice de masse corporelle, le tabagisme ou la région de résidence. L’objectif est donc de développer des modèles capables de traduire ces informations en estimations précises des coûts, afin d’améliorer la gestion des risques et la prise de décision dans le domaine de l’assurance santé.

## 📝 Description
Ce dataset contient des informations sur des bénéficiaires d’assurance santé aux États‑Unis, avec pour objectif de prédire les **coûts médicaux individuels** facturés par l’assurance.

## 📂 Structure du dataset
- **Nombre d’observations** : 1 338 lignes  
- **Nombre de variables** : 7 colonnes

### 🔑 Variables
- `age` : Âge du bénéficiaire principal  
- `sex` : Sexe du contractant (male / female)  
- `bmi` : Indice de masse corporelle (kg/m²)  
- `children` : Nombre d’enfants à charge couverts par l’assurance  
- `smoker` : Indique si le bénéficiaire est fumeur (yes / no)  
- `region` : Région de résidence aux États‑Unis (northeast, southeast, southwest, northwest)  
- `charges` : Coûts médicaux individuels facturés par l’assurance (variable cible)

## 🎯 Objectif
Construire des modèles de **régression** pour prédire les charges médicales.

```python
"""
==============================================================================
EXPLORATIVE DATA ANALYSIS (EDA) - ASSURANCE MÉDICALE
==============================================================================
Auteur: Votre équipe
Date: 2024
Description: Analyse exploratoire complète des coûts d'assurance médicale
Variables: age, sex, bmi, children, smoker, region, charges
==============================================================================
"""

# ==============================================================================
# 1. CONFIGURATION ET IMPORT DES BIBLIOTHÈQUES
# ==============================================================================

import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns
import warnings
from scipy import stats
from scipy.stats import pearsonr, f_oneway, kruskal

# Configuration
warnings.filterwarnings('ignore')
plt.style.use('seaborn-v0_8-darkgrid')
sns.set_palette("husl")

# Paramètres de visualisation
plt.rcParams['figure.figsize'] = (12, 6)
plt.rcParams['font.size'] = 10
plt.rcParams['axes.titlesize'] = 14
plt.rcParams['axes.labelsize'] = 12
plt.rcParams['xtick.labelsize'] = 10
plt.rcParams['ytick.labelsize'] = 10

# Couleurs personnalisées
COLORS = {
    'primary': '#2E86AB',
    'secondary': '#A23B72',
    'accent': '#F18F01',
    'success': '#06A77D',
    'warning': '#D00000'
}

print("=" * 80)
print(" " * 20 + "EXPLORATIVE DATA ANALYSIS")
print(" " * 15 + "Analyse des Coûts d'Assurance Médicale")
print("=" * 80)
print()

# ==============================================================================
# 2. COPIE ET PRÉPARATION DES DONNÉES
# ==============================================================================

# Créer une copie pour éviter de modifier le dataframe original
#df = data.copy()

print("✓ Données chargées et copiées avec succès!")
print()

# ==============================================================================
# 3. VUE D'ENSEMBLE DES DONNÉES
# ==============================================================================

print("\n" + "=" * 80)
print("3. VUE D'ENSEMBLE DES DONNÉES")
print("=" * 80 + "\n")

# 3.1 Dimensions du dataset
print(f"📊 Dimensions du dataset: {df.shape[0]} lignes × {df.shape[1]} colonnes\n")

# 3.2 Aperçu des premières lignes
print("📋 Aperçu des 10 premières lignes:")
print(df.head(10))
print()

# 3.3 Informations sur les colonnes
print("📝 Informations sur les colonnes:")
print(df.info())
print()

# 3.4 Séparation des variables par type
numerical_vars = df.select_dtypes(include=[np.number]).columns.tolist()
categorical_vars = df.select_dtypes(include=['object', 'category']).columns.tolist()

print(f"\n🔢 Variables Numériques ({len(numerical_vars)}):")
for var in numerical_vars:
    print(f"   • {var}")

print(f"\n📊 Variables Catégorielles ({len(categorical_vars)}):")
for var in categorical_vars:
    print(f"   • {var}")

# 3.5 Identification de la variable cible
target_var = 'charges'
numerical_features = [col for col in numerical_vars if col != target_var]

print(f"\n🎯 Variable Cible: {target_var}")
print(f"📐 Variables Prédictives Numériques ({len(numerical_features)}): {numerical_features}")
print(f"📋 Variables Prédictives Catégorielles ({len(categorical_vars)}): {categorical_vars}")

# 3.6 Aperçu des types de données détaillé
print("\n" + "-" * 80)
print("Résumé des types de données:")
print("-" * 80)
type_summary = pd.DataFrame({
    'Variable': df.columns,
    'Type': df.dtypes.values,
    'Non-Null Count': df.count().values,
    'Null Count': df.isnull().sum().values,
    'Unique Values': df.nunique().values
})
print(type_summary.to_string(index=False))
print()

# 3.7 Visualisation de la structure des données
fig, axes = plt.subplots(1, 2, figsize=(14, 5))

# Graphique 1: Types de variables
type_counts = pd.Series({
    'Numériques': len(numerical_vars),
    'Catégorielles': len(categorical_vars)
})
colors_types = [COLORS['primary'], COLORS['secondary']]
wedges, texts, autotexts = axes[0].pie(
    type_counts,
    labels=type_counts.index,
    autopct='%1.1f%%',
    startangle=90,
    colors=colors_types,
    textprops={'fontsize': 12, 'fontweight': 'bold'}
)
for autotext in autotexts:
    autotext.set_color('white')
axes[0].set_title('📊 Répartition des Types de Variables',
                   fontsize=14, fontweight='bold', pad=20)

# Graphique 2: Nombre de valeurs uniques par variable
unique_counts = df.nunique().sort_values(ascending=True)
bars = axes[1].barh(unique_counts.index, unique_counts.values,
                     color=COLORS['accent'], edgecolor='black', alpha=0.8)
axes[1].set_xlabel('Nombre de Valeurs Uniques', fontweight='bold')
axes[1].set_ylabel('Variables', fontweight='bold')
axes[1].set_title('🔢 Cardinalité des Variables', fontsize=14, fontweight='bold', pad=20)
axes[1].grid(axis='x', alpha=0.3)

# Ajouter les valeurs sur les barres
for i, (bar, val) in enumerate(zip(bars, unique_counts.values)):
    axes[1].text(val + max(unique_counts.values)*0.01, i, str(val),
                 va='center', fontweight='bold')

plt.suptitle('🔍 Structure du Dataset', fontsize=16, fontweight='bold', y=1.02)
plt.tight_layout()
plt.show()

print("\n✓ Section 3 terminée: Vue d'ensemble des données complète")
print("=" * 80)
```
